## Exploratory Data Analysis(EDA)

# 1.Website Traffic Source Analysis by users

```python
df.groupby("Session primary channel group (Default channel group)")["Users"].sum().sort_values(ascending=False)
```

```python
Traffic=df.groupby("Session primary channel group (Default channel group)")["Users"].sum().sort_values(ascending=False)
```

```python
Traffic.plot(kind="bar",figsize=(8,5))
plt.title("Total Users By Traffic Sorce")
plt.xlabel("Traffic Source")
plt.ylabel("Total Users")
plt.xticks(rotation=45)
plt.show()
```

## Buisiness Insights

## Organic Social is the top-performing traffic sources with 47,572 users,indicating social media is the strongest channel for attracting visitors.The company should continue investing in social media marketing.Direct,Organic Search and Referral also contribute a significant number of users.Email traffic is extremely low( only 2 users).Company should strengthern email campaigns and other underperforming channels.
    

## 2.Website Traffic Source Analysis By Sessions

```python
df.groupby("Session primary channel group (Default channel group)")["Sessions"].sum().sort_values(ascending=False)
```

```python
session=df.groupby("Session primary channel group (Default channel group)")["Sessions"].sum().sort_values(ascending=False)
```

```python
session.plot(kind="bar",figsize=(8,5))
plt.title("Total Sessions By Traffic Sorce")
plt.xlabel("Traffic Source")
plt.ylabel("Total Sessions")
plt.xticks(rotation=45)
plt.show()
```

## Buisiness Insights

## Organic Social is the leading traffic source with 60,627 sessions, followed by Direct and Organic Search. This shows that social media is the primary driver of website visits, while Email and Organic Video require improvement to increase user traffic.

## 3.Website Traffic Source Analysis by Engagement Rate

```python
df.groupby("Session primary channel group (Default channel group)")["Engagement rate"].mean().sort_values(ascending=False)
```

```python
engagement=df.groupby("Session primary channel group (Default channel group)")["Engagement rate"].mean().sort_values(ascending=False)
```

```python
engagement.plot(kind="bar",figsize=(8,5))
plt.title("Average Engagment Rate By Traffic Sorce")
plt.xlabel("Traffic Source")
plt.ylabel("Average Engagment Rate")
plt.xticks(rotation=45)
plt.show()
```

## Buisiness Insights

## Organic Video recorded the highest engagement rate (76%), indicating the highest quality of user engagement. Referral and Organic Search also performed well, while Organic Social had a comparatively lower engagement rate. The business should focus on improving engagement for high-traffic channels while maintaining the performance of high-engagement channels.

## 5. Average Session Duration by Traffic Source

```python
df.groupby("Session primary channel group (Default channel group)")["Average engagement time per session"].mean().sort_values(ascending=False)
```

```python
session_duration=df.groupby("Session primary channel group (Default channel group)")["Average engagement time per session"].mean().sort_values(ascending=False)
```

```python
session_duration.plot(kind="bar",figsize=(8,5))
plt.title("Average Session Duration By Traffic Sources")
plt.xlabel("Traffic Source")
plt.ylabel("Average Session Duration")
plt.xticks(rotation=45)
plt.show()
```

## Buisiness Insights

## Organic Video recorded the highest average session duration (180.36), showing that visitors from this channel spend the most time on the website. Referral also performs well, while Direct (45.53) and Organic Search (47.01) have the lowest session duration, indicating an opportunity to improve user engagement on these channels.

## 6.Events per Session by Traffic Sources

## 7.Website Traffic Trend Over Time

```python
df.groupby("Date + hour (YYYYMMDDHH)")["Sessions"].sum().reset_index()
```

```python
df["Date"]=df["Date + hour (YYYYMMDDHH)"].dt.date
```

```python
traffic_trend=df.groupby("Date")["Sessions"].sum().reset_index()
```

```python
traffic_trend.plot(
    x="Date",
    y="Sessions",
    kind="line",
    figsize=(8,5)
)
plt.title("Website Traffic Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Total Sessions")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

## Buisiness Insight

## Website sessions fluctuated across the analysis period, with noticeable peaks and dips in daily traffic.
